```python
import heapq

class Node:
    def __init__(self, state, cost, path):
        self.state = state
        self.cost = cost
        self.path = path

    def __lt__(self, other):
        return self.cost < other.cost

def uniform_cost_search(graph, start, goal):
    priority_queue = [(0, start, [start])]

    # Set to keep track of visited nodes to avoid cycles and redundant processing
    visited = set()

    # Dictionary to store the minimum cost found so far to reach a state
    min_cost = {start: 0}

    while priority_queue:
        current_cost, current_state, current_path = heapq.heappop(priority_queue)

        if current_state == goal:
            return current_cost, current_path

        if current_state in visited and current_cost >= min_cost.get(current_state, float('inf')):
            continue # Skip if we found a more expensive path to an already visited node

        visited.add(current_state)

        for neighbor, edge_cost in graph.get(current_state, {}).items():
            new_cost = current_cost + edge_cost
            if neighbor not in visited or new_cost < min_cost.get(neighbor, float('inf')):
                min_cost[neighbor] = new_cost
                new_path = current_path + [neighbor]
                heapq.heappush(priority_queue, (new_cost, neighbor, new_path))

    return None, None # No path found


graph = {
    'A': {'B': 1, 'C': 10},
    'B': {'D': 2, 'C': 5},
    'C': {'E': 3},
    'D': {'E': 8},
    'E': {}
}

start_node = 'A'
goal_node = 'E'

cost, path = uniform_cost_search(graph, start_node, goal_node)

if path:
    print(f"Lowest cost path from {start_node} to {goal_node}: {path}")
    print(f"Total cost: {cost}")
else:
    print(f"No path found from {start_node} to {goal_node}")

print("\n--- Another Example ---")
graph2 = {
    'S': {'A': 1, 'G': 10},
    'A': {'B': 5},
    'B': {'G': 1},
    'G': {}
}

start_node2 = 'S'
goal_node2 = 'G'

cost2, path2 = uniform_cost_search(graph2, start_node2, goal_node2)

if path2:
    print(f"Lowest cost path from {start_node2} to {goal_node2}: {path2}")
    print(f"Total cost: {cost2}")
else:
    print(f"No path found from {start_node2} to {goal_node2}")
```
